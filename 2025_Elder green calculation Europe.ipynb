```python
import yfinance as yf
import pandas as pd
import numpy as np

# List of tickers
my_tickers =  ["ADS.DE", "AIR.DE", "ALV.DE", "BAS.DE", "BAYN.DE", "BEI.DE", "BMW.DE", "CON.DE",
              "1COV.DE", "DHER.DE", "DTE.DE", "DB1.DE", "DBK.DE", "DHL.DE", "DWS.DE", "EOAN.DE",
              "FME.DE", "FRE.DE", "HEN3.DE", "HEI.DE", "IFX.DE", "LIN.DE", "MRK.DE", "MTX.DE",
              "MUV2.DE", "PAH3.DE", "PUM.DE", "QIA.DE", "RWE.DE", "SAP.DE", "SIE.DE", "SY1.DE",
              "SHL.DE", "VNA.DE", "VOW3.DE", "ZAL.DE", "AI.PA", "AIR.PA", "BNP.PA", "EN.PA", "GLE.PA", "MC.PA", "OR.PA", "RI.PA", "SAN.PA", "SGO.PA",
              "STM.PA", "SU.PA", "VIE.PA", "DG.PA", "LR.PA", "CA.PA", "VIV.PA", "URW.PA", "HO.PA", "CAP.PA",
              "ATO.PA", "PUB.PA", "EL.PA", "KER.PA", "RMS.PA", "ACA.PA", "ALO.PA", "CS.PA", "EDF.PA", "ML.PA",
              "RNO.PA", "SW.PA", "ENGI.PA", "SGO.PA", "MT.PA", "FP.PA","AAL.L", "ABF.L", "ADM.L", "AHT.L", "ANTO.L", "AV.L", "AZN.L", "BA.L", "BARC.L", "BATS.L",
                "BDEV.L", "BLND.L", "BME.L", "BP.L", "BT-A.L", "CCL.L", "CPG.L", "CRH.L", "DGE.L", "ENT.L",
                "EZJ.L", "GSK.L", "HSBA.L", "IAG.L", "ICP.L", "IHG.L", "III.L", "IMB.L", "ITV.L", "KGF.L",
                "LGEN.L", "LLOY.L", "MNG.L", "NG.L", "NXT.L", "OCDO.L", "PRU.L", "PSN.L", "RB.L", "REL.L",
                "RIO.L", "RR.L", "RTO.L", "SBRY.L", "SDR.L", "SGE.L", "SHEL.L", "SMDS.L", "SMIN.L", "STAN.L",
                "TSCO.L", "ULVR.L", "VOD.L", "WTB.L"]

# Function to calculate EMA and MACD histogram
def calc_hist(source, fast_length, slow_length, signal_length):
    fast_ema = source.ewm(span=fast_length, adjust=False).mean()
    slow_ema = source.ewm(span=slow_length, adjust=False).mean()
    macd = fast_ema - slow_ema
    signal = macd.ewm(span=signal_length, adjust=False).mean()
    hist = macd - signal
    return hist, fast_ema

# Function to determine candle color based on Elder Impulse System
def get_color(ema_series, macd_hist):
    return np.where((ema_series > ema_series.shift(1)) & (macd_hist > macd_hist.shift(1)), 'green', 
                    np.where((ema_series < ema_series.shift(1)) & (macd_hist < macd_hist.shift(1)), 'red', 'blue'))

# Parameters
length_ema = 13
fast_length = 12
slow_length = 26
signal_length = 9

# Dictionary to store results
stock_day_counts = []

# Loop through tickers
for symbol in my_tickers:
    try:
        print(f"Processing {symbol}...")
        
        # Download data
        data = yf.download(symbol, start="2024-03-13", end="2025-03-13", progress=False)
        if data.empty:
            continue
        
        # Calculate MACD histogram and EMA
        macd_hist, fast_ema = calc_hist(data['Close'], fast_length, slow_length, signal_length)
        
        # Determine candle colors
        bar_colors = get_color(fast_ema, macd_hist)
        
        # Count green, blue, and red days
        green_days = np.sum(bar_colors == 'green')
        blue_days = np.sum(bar_colors == 'blue')
        red_days = np.sum(bar_colors == 'red')
        
        # Save results
        stock_day_counts.append([symbol, green_days, blue_days, red_days])
    
    except Exception as e:
        print(f"Error processing {symbol}: {e}")

# Convert to DataFrame and save as CSV
df_results = pd.DataFrame(stock_day_counts, columns=['Ticker', 'Green Days', 'Blue Days', 'Red Days'])
df_results.to_csv("stock_analysis_results.csv", index=False, encoding='utf-8')

print("Results saved to stock_analysis_results.csv")
```
